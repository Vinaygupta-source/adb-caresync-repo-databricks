```python
dbutils.widgets.text("landing_timestamp", "", "Landing Timestamp")
dbutils.widgets.text("table_name", "", "Table Name")

landing_timestamp = dbutils.widgets.get("landing_timestamp")
table_name = dbutils.widgets.get("table_name")
```

```python
bronze_df = spark.read.table(f"adb_caresync_7405619098016345.bronze.{table_name}").filter(f"landing_timestamp='{landing_timestamp}'")
```

```python
from pyspark.sql.functions import lit, current_timestamp

updated_df = bronze_df.withColumn("insert_timestamp",current_timestamp()).withColumn("update_timestamp",current_timestamp()).drop("ingest_timestamp")
```

```python
updated_df.write.mode("overwrite").saveAsTable(f"adb_caresync_7405619098016345.silver.{table_name}")
```
